Consider the task: **comparing values across categories**.

Ranking (most to least effective): Position, length, angle, area/size, volume, shape, color hue, color lightness

GenAI response:
1. Position (Most Effective)
From a perceptual standpoint, our visual system is highly attuned to detecting spatial differences. The brain processes position through the parietal cortex, which specializes in spatial relationships. When comparing positions on a common scale, we use a simple mental operation: measuring distance from a baseline. This requires minimal cognitive load because we don't need to perform any mental transformations. Research shows we can detect positional differences as small as 1-2% of the total scale.
2. Length (Highly Effective)
Length comparison relies on our ability to perceive extent, which is processed early in the visual system. When bars are aligned to a common baseline, comparing lengths becomes a position judgment (comparing endpoints). However, when lengths are not aligned, accuracy drops significantly because our brains must mentally "move" one object to compare it. This additional cognitive step introduces error. This explains why stacked bar charts are harder to read than grouped bar charts.
3. Area (Moderately Effective)
Our perception of area is systematically biased. Studies by Stevens found that perceived area scales as a power function of actual area, with an exponent around 0.7. This means we consistently underestimate differences in area, especially for larger shapes. For circles, we often misjudge because we unconsciously use diameter as a proxy for area. Cognitively, area requires us to integrate two dimensions simultaneously, which demands more processing than one-dimensional judgments like length.
4. Color Hue (Least Effective for Quantities)
Color hue is processed categorically, not continuously. Our brains evolved to distinguish distinct colors (ripe vs unripe fruit, for example) rather than to perceive ordered gradations. There is no natural ordering to hues—red is not "more than" blue in any intuitive sense. This makes hue ineffective for representing quantities. Additionally, approximately 8% of males have some form of color vision deficiency, making hue-based comparisons inaccessible. Hue works well for categorical distinctions because our visual system can quickly identify "different" colors preattentively.
